```python
import pandas as pd
import numpy as np

print("Libraries imported successfully.")
```

## 2. Load the Raw Dataset

The original dataset is kept unchanged.

We will load it into a Pandas DataFrame called `df`.

```python
df = pd.read_csv("raw_dataset.csv")

print("Dataset loaded successfully.")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

- Data types

## 6. List All Columns

The raw dataset contains demographic information, monthly expenses, savings information, and potential savings variables.

```python
print("Columns in the raw dataset:\n")

for i, column in enumerate(df.columns, start=1):
    print(f"{i}. {column}")
```

## 7. Check Missing Values

Missing values can affect machine learning models.

We therefore check the number of missing values in every column.

```python
missing_values = df.isnull().sum()

missing_table = pd.DataFrame({
    "Column": missing_values.index,
    "Missing_Values": missing_values.values
})

missing_table
```

```python
print("Total missing values in the dataset:", df.isnull().sum().sum())
```

## 8. Check Duplicate Records

Duplicate records can cause certain users or observations to be overrepresented.

We check for exact duplicate rows.

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

## 9. Examine Age Distribution

SpendWise is designed for young adults.

Our project requirement is:

**Age between 18 and 35 years, inclusive.**

```python
print("Minimum age:", df["Age"].min())
print("Maximum age:", df["Age"].max())

print("\nAge statistics:")
print(df["Age"].describe())
```

```python
age_groups = pd.cut(
    df["Age"],
    bins=[0, 17, 25, 30, 35, 40, 50, 65, 100],
    labels=[
        "Below 18",
        "18-25",
        "26-30",
        "31-35",
        "36-40",
        "41-50",
        "51-65",
        "Above 65"
    ],
    right=True
)

age_distribution = age_groups.value_counts().sort_index()

age_distribution.to_frame(name="Number_of_Records")
```

## 10. Filter the Target Age Group

The target population for SpendWise is young adults aged **18–35**.

Therefore, records outside this age range will be removed.

We use an inclusive range:

**18 ≤ Age ≤ 35**

```python
original_row_count = len(df)

df = df[
    (df["Age"] >= 18) &
    (df["Age"] <= 35)
].copy()

filtered_row_count = len(df)

removed_rows = original_row_count - filtered_row_count

print("Rows before age filtering:", original_row_count)
print("Rows after age filtering:", filtered_row_count)
print("Rows removed:", removed_rows)
```

```python
print("Age range after filtering:")
print("Minimum age:", df["Age"].min())
print("Maximum age:", df["Age"].max())
```

## 11. Examine Occupation Categories

The raw dataset contains four occupation categories.

Our project requires three categories:

- Student
- Salaried
- Self-employed/Business

The original category `Professional` will be mapped to `Salaried`.

`Retired` will be removed because it is outside our defined project population.

```python
occupation_table = df["Occupation"].value_counts().reset_index()

occupation_table.columns = [
    "Occupation",
    "Number_of_Records"
]

occupation_table
```

## 12. Remove the Retired Category

The `Retired` category is not part of our target occupation groups.

We therefore retain only:

- Student
- Professional
- Self_Employed

```python
before_occupation_filter = len(df)

allowed_occupations = [
    "Student",
    "Professional",
    "Self_Employed"
]

df = df[
    df["Occupation"].isin(allowed_occupations)
].copy()

after_occupation_filter = len(df)

print("Rows before occupation filtering:", before_occupation_filter)
print("Rows after occupation filtering:", after_occupation_filter)
print("Rows removed:", before_occupation_filter - after_occupation_filter)
```

## 13. Standardize Occupation Names

The raw dataset uses `Professional` and `Self_Employed`.

For SpendWise, we use clearer project terminology:

- `Professional` → `Salaried`
- `Self_Employed` → `Self-employed/Business`
- `Student` → `Student`

```python
occupation_mapping = {
    "Student": "Student",
    "Professional": "Salaried",
    "Self_Employed": "Self-employed/Business"
}

df["Occupation"] = df["Occupation"].replace(
    occupation_mapping
)

print(df["Occupation"].value_counts())
```

## 14. Standardize City Tier

The raw dataset stores city tiers as:

- Tier_1
- Tier_2
- Tier_3

We standardize the naming format to:

- Tier 1
- Tier 2
- Tier 3

```python
df["City_Tier"] = df["City_Tier"].str.replace(
    "_",
    " ",
    regex=False
)

city_tier_table = df["City_Tier"].value_counts().reset_index()

city_tier_table.columns = [
    "City_Tier",
    "Number_of_Records"
]

city_tier_table
```

## 15. Remove Unused Dependents Feature

The `Dependents` column is not required for the current scope of SpendWise.

It will therefore be removed from the cleaned dataset.

This removes the column only; no records are removed because of this feature.

```python
df = df.drop(columns=["Dependents"])

print("Dependents column removed.")
print("Number of columns:", len(df.columns))
```

## 16. Validate Income Values

Income is an important feature for SpendWise.

We check for:
- Minimum income
- Maximum income
- Zero income
- Negative income
- Statistical distribution

```python
print("Minimum income:", df["Income"].min())
print("Maximum income:", df["Income"].max())

print("\nZero income records:", (df["Income"] == 0).sum())
print("Negative income records:", (df["Income"] < 0).sum())

print("\nIncome statistics:")
df["Income"].describe()
```

## 17. Validate Expense Values

We check whether the major expense columns contain negative values.

Negative monthly expenses are not meaningful for our application and would indicate invalid records.

```python
expense_columns = [
    "Rent",
    "Loan_Repayment",
    "Insurance",
    "Groceries",
    "Transport",
    "Eating_Out",
    "Entertainment",
    "Utilities",
    "Healthcare",
    "Education",
    "Miscellaneous"
]

negative_expenses = (df[expense_columns] < 0).sum()

negative_expense_table = negative_expenses.reset_index()

negative_expense_table.columns = [
    "Expense_Column",
    "Negative_Values"
]

negative_expense_table
```

## 18. Create the Food Expense Category

SpendWise uses Food as one of its primary categories.

The raw dataset contains:

- Groceries
- Eating_Out

Therefore:

**Food Expense = Groceries + Eating Out**

```python
df["Food_Expense"] = (
    df["Groceries"] +
    df["Eating_Out"]
)

df[
    [
        "Groceries",
        "Eating_Out",
        "Food_Expense"
    ]
].head()
```

## 19. Create the Travel Expense Category

The raw dataset contains `Transport`.

For SpendWise:

**Travel Expense = Transport**

```python
df["Travel_Expense"] = df["Transport"]

df[
    [
        "Transport",
        "Travel_Expense"
    ]
].head()
```

## 20. Create the Entertainment Expense Category

The raw dataset already contains an `Entertainment` expense.

For consistency, we create:

**Entertainment Expense = Entertainment**

```python
df["Entertainment_Expense"] = df["Entertainment"]

df[
    [
        "Entertainment",
        "Entertainment_Expense"
    ]
].head()
```

## 21. Review the Three Main SpendWise Expense Categories

SpendWise focuses on three primary expense categories:

1. Food
2. Travel
3. Entertainment

These categories are used to analyze the user's spending
behavior and will form the basis for future expense prediction.

```python
category_columns = [
    "Food_Expense",
    "Travel_Expense",
    "Entertainment_Expense"
]

df[category_columns].head()
```

## 22. Calculate Total SpendWise Expense

Total SpendWise Expense represents the user's total spending
across the three expense categories included in the project.

Total Expense = Food + Travel + Entertainment

```python
df["Total_Expense"] = (
    df["Food_Expense"]
    + df["Travel_Expense"]
    + df["Entertainment_Expense"]
)

df[
    [
        "Food_Expense",
        "Travel_Expense",
        "Entertainment_Expense",
        "Total_Expense"
    ]
].head()
```

## 23. Calculate Actual Savings

Actual Savings represents the income remaining after
the user's SpendWise expenses.

Actual Savings = Income - Total Expense

```python
df["Remaining_Income"] = (
    df["Income"] - df["Total_Expense"]
)

df[
    ["Income", "Total_Expense", "Remaining_Income"]
].head()
```

## 24. Calculate Savings Rate

Savings Rate represents the percentage of income remaining
after the user's SpendWise expenses.

Savings Rate = (Actual Savings / Income) × 100

```python
df["Remaining_Income_Percentage"] = (
    df["Remaining_Income"] / df["Income"]
) * 100

df[
    ["Income", "Remaining_Income", "Remaining_Income_Percentage"]
].head()
```

## 25. Calculate Category-wise Spending Percentages

These percentages show how the user's SpendWise expenses
are distributed across Food, Travel, and Entertainment.

The three percentages together represent the user's
spending distribution across the project's primary categories.

```python
df["Food_Percentage"] = (
    df["Food_Expense"] / df["Total_Expense"]
) * 100

df["Travel_Percentage"] = (
    df["Travel_Expense"] / df["Total_Expense"]
) * 100

df["Entertainment_Percentage"] = (
    df["Entertainment_Expense"] / df["Total_Expense"]
) * 100

df[
    [
        "Food_Percentage",
        "Travel_Percentage",
        "Entertainment_Percentage"
    ]
].head()
```

## 26. Identify Overspending Users

An overspending flag is created when the user's total
SpendWise expense is greater than their income.

Overspending Flag:
- 1 = Total Expense > Income
- 0 = Total Expense <= Income

This feature can later be used to generate spending
warnings and financial alerts.

```python
df["Overspending_Flag"] = (
    df["Total_Expense"] > df["Income"]
).astype(int)

print(
    "Number of overspending records:",
    df["Overspending_Flag"].sum()
)

print(
    "Overspending percentage:",
    round(
        df["Overspending_Flag"].mean() * 100,
        2
    ),
    "%"
)
```

```python
df = df.drop(
    columns=["Overspending_Flag"],
    errors="ignore"
)

print("Overspending_Flag removed.")
```

## 27. Remove Expense Categories Outside the SpendWise Scope

The original dataset contains several additional financial
expense categories such as:

- Rent
- Loan Repayment
- Insurance
- Healthcare
- Education
- Utilities
- Miscellaneous

These categories are outside the primary scope of the
current SpendWise project.

SpendWise focuses specifically on:

- Food
- Travel
- Entertainment

Therefore, these additional expense columns are removed
from the cleaned modeling dataset.

```python
excluded_expense_columns = [
    "Rent",
    "Loan_Repayment",
    "Insurance",
    "Utilities",
    "Healthcare",
    "Education",
    "Miscellaneous"
]

df = df.drop(
    columns=excluded_expense_columns,
    errors="ignore"
)

print("Excluded expense columns removed:")

for column in excluded_expense_columns:
    print("-", column)

print("\nCurrent number of columns:", len(df.columns))
```

## 28. Remove Derived Financial Columns

The original dataset contains Disposable Income and Potential
Savings columns.

These values are derived from other financial information
and may introduce data leakage into the machine learning model.

Therefore, they are removed from the cleaned modeling dataset.

```python
derived_columns = [
    "Disposable_Income",
    "Potential_Savings_Groceries",
    "Potential_Savings_Transport",
    "Potential_Savings_Eating_Out",
    "Potential_Savings_Entertainment",
    "Potential_Savings_Utilities",
    "Potential_Savings_Healthcare",
    "Potential_Savings_Education",
    "Potential_Savings_Miscellaneous"
]

df = df.drop(
    columns=derived_columns,
    errors="ignore"
)

print("Derived financial columns removed.")

print("\nCurrent number of columns:", len(df.columns))
```

## 29. Final SpendWise Dataset Structure

```python
print("Final number of rows:", df.shape[0])
print("Final number of columns:", df.shape[1])

print("\nFinal columns:")

for i, column in enumerate(df.columns, start=1):
    print(f"{i}. {column}")
```

## 30. Final Data Quality Validation

```python
remaining_expense_columns = [
    "Groceries",
    "Eating_Out",
    "Transport",
    "Entertainment"
]

print("========== FINAL DATA QUALITY REPORT ==========\n")

print("Number of rows:", len(df))
print("Number of columns:", len(df.columns))

print("\nMissing values:",
      df.isnull().sum().sum())

print("Duplicate rows:",
      df.duplicated().sum())

print(
    "\nAge range:",
    df["Age"].min(),
    "to",
    df["Age"].max()
)

print("\nOccupation categories:")
print(df["Occupation"].value_counts())

print("\nCity Tier categories:")
print(df["City_Tier"].value_counts())

print(
    "\nNegative expense values:",
    (df[remaining_expense_columns] < 0).sum().sum()
)

print(
    "\nIncome <= 0 records:",
    (df["Income"] <= 0).sum()
)

print("\n==============================================")
```
